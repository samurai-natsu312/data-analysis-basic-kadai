```python
!pip install -q japanize-matplotlib

import matplotlib.pyplot as plt
import numpy as np
import japanize_matplotlib

# 6個のグラフを表示する領域を作成
fig, axes = plt.subplots(2, 3, tight_layout=True)

# 各領域にグラフを描画する
# (0, 0) 折れ線グラフ
y0 = [1, 2, -5, 2]
axes[0, 0].plot(y0)
axes[0, 0].set_title("折れ線グラフ")

# (0, 1) sin関数
x = np.linspace(0, 10, 50)
axes[0, 1].plot(np.sin(x))
axes[0, 1].set_title("sin関数")

# (0, 2) ヒストグラム
data = np.random.randn(100)
axes[0, 2].hist(data, bins=15)
axes[0, 2].set_title("ヒストグラム")

# (1, 0) 散布図
x_scat = np.random.rand(30)
y_scat = np.random.rand(30)
axes[1, 0].scatter(x_scat, y_scat)
axes[1, 0].set_title("散布図")

# グラフを表示させる
plt.show()
```
